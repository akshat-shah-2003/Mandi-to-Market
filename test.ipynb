```python
import requests
import time
import os
from dotenv import load_dotenv
```

```python
load_dotenv()

url = os.getenv("DATA_URL")

params = {
    "api-key": os.getenv("DATASET_API_KEY"),
    "format": "json",
    "limit": 10,
    "filters[state.keyword]": "Madhya Pradesh",
}

headers = {
    "User-Agent": "Mandi-to-Market/1.0"
}

for attempt in range(3):
    response = requests.get(
        url,
        params=params,
        headers=headers,
        timeout=30
    )

    if response.status_code == 200:
        data = response.json()
        break

    print(f"Attempt {attempt + 1} failed: {response.status_code}")
    time.sleep(2)

print(data)
```
